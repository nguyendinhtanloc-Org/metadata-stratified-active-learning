```python
# Test loader functions
from data.loader import labels_to_yolo_lines, BDD100K_TO_YOLO, YOLO_CLASS_NAMES

print("BDD100K → YOLO Mapping:")
for k, v in BDD100K_TO_YOLO.items():
    print(f"  {k} → {v} ({YOLO_CLASS_NAMES[v]})")
```

```python
# Load sample metadata
import json

METADATA_PATH = "/kaggle/input/datasets/solesensei/solesensei_bdd100k/bdd100k_labels_release/bdd100k/labels/bdd100k_labels_images_train.json"

with open(METADATA_PATH, 'r') as f:
    metadata = json.load(f)

sample = metadata[0]
print(f"Sample: {sample['name']}")
print(f"Labels count: {len(sample['labels'])}")
```

```python
# Test labels conversion
yolo_lines = labels_to_yolo_lines(sample['labels'])

print("Original labels:")
for label in sample['labels']:
    cat = label.get('category', 'N/A')
    box2d = label.get('box2d', 'N/A')
    print(f"  {cat}: {box2d}")

print("\nConverted YOLO format:")
for line in yolo_lines:
    print(f"  {line}")
```

```python
# Test với nhiều samples
print("Testing 10 samples...")

total_labels = 0
converted_labels = 0

for i in range(10):
    sample = metadata[i]
    original = sample['labels']
    converted = labels_to_yolo_lines(original)
    
    total_labels += len(original)
    converted_labels += len(converted)
    
    print(f"Sample {i}: {len(original)} labels → {len(converted)} YOLO lines")

print(f"\nTotal: {total_labels} → {converted_labels} converted")
print(f"Conversion rate: {converted_labels/total_labels*100:.1f}%")
```

```python
# Kiểm tra xem ảnh có tồn tại không
import os

IMAGE_DIR = "/kaggle/input/datasets/solesensei/solesensei_bdd100k/bdd100k/bdd100k/images/100k/train/trainA"

print("Checking 10 images exist...")
exists_count = 0
for i in range(10):
    filename = metadata[i]['name']
    path = os.path.join(IMAGE_DIR, filename)
    exists = os.path.exists(path)
    if exists:
        exists_count += 1
    status = "✅" if exists else "❌"
    print(f"  {status} {filename}")

print(f"\nImages found: {exists_count}/10")
```
