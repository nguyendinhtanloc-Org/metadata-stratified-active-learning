```python
# Cấu hình
from active_learning.loop import ActiveLearningLoop

# Đường dẫn
METADATA_PATH = "/kaggle/input/datasets/solesensei/solesensei_bdd100k/bdd100k_labels_release/bdd100k/labels/bdd100k_labels_images_train.json"

# Tất cả thư mục có thể chứa ảnh
IMAGE_DIRS = [
    "/kaggle/input/datasets/solesensei/solesensei_bdd100k/bdd100k/bdd100k/images/100k/train/trainA",
    "/kaggle/input/datasets/solesensei/solesensei_bdd100k/bdd100k/bdd100k/images/100k/train/trainB",
]

# Quick test config
config = {
    "seed": 42,
    "initial_budget": 100,
    "batch_size": 50,
    "num_rounds": 3,  # Chỉ 3 rounds để test nhanh
    "epochs": 2,      # Chỉ 2 epochs để test nhanh
    "device": 0,
    "num_classes": 10,
    "stratify_fields": ["weather", "scene", "timeofday"],
    "model_name": "yolov8n"
}

print("Config:")
for k, v in config.items():
    print(f"  {k}: {v}")
```

```python
# Khởi tạo AL Loop
experiment_name = "quick_test_stratified"

al_loop = ActiveLearningLoop(
    config=config,
    experiment_name=experiment_name
)

print("Active Learning Loop initialized!")
```

```python
# Chạy experiment
print("=" * 60)
print("Starting Quick Test Experiment")
print("=" * 60)

results_df = al_loop.run(
    metadata_path=METADATA_PATH,
    image_dir=IMAGE_DIRS
)
```

```python
# Kết quả
print("\n" + "=" * 60)
print("RESULTS:")
print("=" * 60)
print(results_df)
```

```python
# Visualize
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(results_df['round'], results_df['map50'], marker='o', label='mAP@50')
ax.set_xlabel('Round')
ax.set_ylabel('mAP@50')
ax.set_title('Active Learning Progress - Quick Test')
ax.grid(True)
ax.legend()

plt.tight_layout()
plt.show()
```
